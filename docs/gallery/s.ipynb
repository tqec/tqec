# S

This notebook shows the construction of the logical S (phase) gate with lattice surgery.

## Construction

The logical S gate performs a $\pi/2$ rotation around the $Z$ axis of the Bloch sphere. On Pauli operators it acts as $Z \to Z$ and $X \to Y$. Since $Y = X \cdot Z$, the $X \to Y$ mapping needs a source of the $Z$ component; this is provided by a $Y$-cube ancilla.

The block graph is built from one `XZX` cube on the main qubit worldline between the two ports, plus a side arm at $z = 1$ that connects the trunk to a $Y$-cube. The side arm has another `XZX` cube in the middle and the $Y$-cube at its far end. The $Y$-cube injects the $+i$ phase that turns $X$ into $Y$.

`tqec` provides a builtin function `tqec.gallery.s` to construct the logical S gate.

```python
from tqec.gallery import s

graph = s()
graph.view_as_html()
```

The logical S has two independent stabilizer flow generators:
* `Z -> Z`
* `X -> Y`

Here we show the correlation surfaces for the generators. The `X -> Y` surface visibly detours into the $Y$-cube ancilla: the surface has an $X$-flavoured piece on the trunk and a $Z$-flavoured piece on the ancilla arm, and together they encode $Y = X \cdot Z$ at the output.

```python
correlation_surfaces = graph.find_correlation_surfaces()
stab_to_surface = {s.external_stabilizer_on_graph(graph): s for s in correlation_surfaces}
print(stab_to_surface.keys())
```

#### `Z -> Z`

```python
graph.view_as_html(
    pop_faces_at_directions=("-Y",),
    show_correlation_surface=stab_to_surface["ZZ"],
)
```

#### `X -> Y`

```python
graph.view_as_html(
    pop_faces_at_directions=("-Y",),
    show_correlation_surface=stab_to_surface["YX"],
)
```
